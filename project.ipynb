**index**: An internal DataFrame index, typically a unique identifier for each row.

**category:** The general product category (e.g., 'Set', 'Kurta').

**size:** The size of the product (e.g., 'S', 'M', 'L').

**product_id:** Amazon Standard Identification Number, a unique identifier for products on Amazon.

**delivery_status:** The status of the package from the courier's perspective (e.g., 'Shipped', 'Unshipped').

**quantity:** The number of units of the product ordered.

**currency:** The currency in which the transaction was made (e.g., 'INR').

**price:** The total monetary value of the order.

**ship_city:** The city where the order was shipped.

**ship_state:** The state where the order was shipped

**ship_postal_code: **The postal code of the shipping destination.

```python
# Rename columns for better clarity

rename_dict = {
    'fulfilment': 'shipped_by',
    'ship-service-level': 'delivery_type',
    'asin': 'product_id',
    'courier_status': 'delivery_status',
    'qty': 'quantity',
    'amount': 'price'
}

df = df.rename(columns=rename_dict)
df.head(2)
```

```python
# Convert all column names to snake_case:snakecaseing

df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
df.columns = df.columns.str.replace('-','_')
df.head()
```

```python
#.copy() ensure my original raw data stays safe and untouched
# Feature Engineering: Creating time-based columns for the dashboard
# Order Month: Extracted from 'Date' to see monthly trends.
# Day of Week: To identify which days (e.g., weekends vs. weekdays) have higher sales.

df_cleaned = df.copy()
df_cleaned['date'] = pd.to_datetime(df_cleaned['date'])
df_cleaned['month'] = df_cleaned['date'].dt.month_name()
df_cleaned['day_of_week'] = df_cleaned['date'].dt.day_name()
df_cleaned.head()
```

```python
# 1. Shipping Region Mapping
state_region_map = {
    'MAHARASHTRA': 'West', 'KARNATAKA': 'South', 'TAMIL NADU': 'South', 'TELANGANA': 'South', 'UTTAR PRADESH': 'North',
    'DELHI': 'North', 'KERALA': 'South', 'WEST BENGAL': 'East', 'ANDHRA PRADESH': 'South', 'GUJARAT': 'West',
    'HARYANA': 'North', 'RAJASTHAN': 'North', 'MADHYA PRADESH': 'Central', 'BIHAR': 'East', 'ODISHA': 'East',
    'PUNJAB': 'North', 'CHATTISGARH': 'Central', 'UTTARAKHAND': 'North', 'JHARKHAND': 'East', 'GOA': 'West',
    'HIMACHAL PRADESH': 'North', 'JAMMU & KASHMIR': 'North', 'ASSAM': 'Northeast', 'MANIPUR': 'Northeast'}

df_cleaned['shipping_region'] = df_cleaned['ship_state'].str.upper().map(state_region_map).fillna('Other')

# 2. Revenue Calculation (Total Amount per SKU)
# We'll create a mapping of SKU to their total sales volume to see business impact

total_sku_revenue = df_cleaned.groupby('sku')['price'].transform('sum')
df_cleaned['sku_revenue'] = total_sku_revenue
df_cleaned.head()
```

```python
# Calculate Q1 (25th percentile) and Q3 (75th percentile) for 'price' grouped by 'category'
Q1 = df_cleaned.groupby('category')['price'].transform(lambda x: x.quantile(0.25))
Q3 = df_cleaned.groupby('category')['price'].transform(lambda x: x.quantile(0.75))

# Calculate the Interquartile Range (IQR)
IQR = Q3 - Q1

# Define upper and lower bounds for outlier detection
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

# Identify outliers: prices outside the calculated bounds for their respective categories
outliers_by_category = df_cleaned[(df_cleaned['price'] < lower_bound) | (df_cleaned['price'] > upper_bound)]

print(f"Number of outliers identified: {len(outliers_by_category)}")
print("Unique categories with outliers:")
print(outliers_by_category['category'].unique())
```

```python
# Define the categories to apply median imputation to
target_categories = ['kurta', 'Western Dress', 'Set', 'Blouse', 'Top', 'Bottom', 'Saree']

# Calculate the median price for each category, ignoring NaN values
category_medians = df_cleaned.groupby('category')['price'].median()

# Identify rows that are in target_categories AND have null prices
mask = (df_cleaned['category'].isin(target_categories)) & (df_cleaned['price'].isnull())

# Fill null 'price' values for these specific rows using the pre-calculated category medians
df_cleaned.loc[mask, 'price'] = df_cleaned.loc[mask, 'category'].map(category_medians)

print("Null values in 'price' after median imputation for target categories:")
print(df_cleaned[df_cleaned['category'].isin(target_categories)]['price'].isnull().sum())
```
